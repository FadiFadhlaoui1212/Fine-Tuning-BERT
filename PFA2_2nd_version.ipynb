```python
# Ouvrir le fichier JSON
with open('/content/split10.json', 'r') as f:
    # Charger le contenu JSON
    testing_data = json.load(f)
```

```python
sentences = []
for i in range (len(data)):
  sentence = []
  for j in range(len(data[i]['tokens'])):
    sentence.append(data[i]['tokens'][j]['token'])
  sentences.append(sentence)
for i in range (len(testing_data)):
  sentence = []
  for j in range(len(testing_data[i]['tokens'])):
    sentence.append(testing_data[i]['tokens'][j]['token'])
  sentences.append(sentence)
```

```python
labels = []
for i in range (len(data)):
  label = []
  for j in range(len(data[i]['tokens'])):
    if not (data[i]['tokens'][j]['tags'][0]['tags'] == None):
      label.append(data[i]['tokens'][j]['tags'][0]['tags'][0]['value'])
    else:
      label.append(data[i]['tokens'][j]['tags'][0]['value'])
  labels.append(label)
for i in range (len(testing_data)):
  label = []
  for j in range(len(testing_data[i]['tokens'])):
    if not (testing_data[i]['tokens'][j]['tags'][0]['tags'] == None):
      label.append(testing_data[i]['tokens'][j]['tags'][0]['tags'][0]['value'])
    else:
      label.append(testing_data[i]['tokens'][j]['tags'][0]['value'])
  labels.append(label)
```

```python
import torch
from transformers import BertTokenizer, BertModel

# Chargez le tokenizer BERT pour l'arabe
tokenizer = BertTokenizer.from_pretrained("asafaya/bert-base-arabic")

# Chargez le modèle BERT pour l'arabe
model = BertModel.from_pretrained("asafaya/bert-base-arabic")


# Convertir les mots en IDs des tokens BERT pour chaque phrase
input_ids = [tokenizer.convert_tokens_to_ids(sentence) for sentence in sentences]

# Obtenez les embeddings de chaque mot
embeddings = []

for ids in input_ids:
    with torch.no_grad():
        outputs = model(torch.tensor(ids).unsqueeze(0))  # Ajouter une dimension batch
    word_embeddings = outputs.last_hidden_state.squeeze(0)  # Retirer la dimension batch
    embeddings.append(word_embeddings)

# Vérifiez la forme des embeddings de chaque mot
#for i, emb in enumerate(embeddings):
#print(f"Forme des embeddings de la phrase {i + 1}:", emb.shape)

```

```python
classes = []
for i in range(len(labels)):
  for j in range(len(labels[i])):
    if not (labels[i][j] in classes):
      classes.append(labels[i][j])
```

```python
def position(value,list):
  for k in range(len(list)):
    if(list[k]==value):
      return k
  return -1
```

```python
X_train = []
Y_train = []
for i in range(len(data)):
  for j in range(len(embeddings[i])):
    X_train.append(embeddings[i][j])
    Y_train.append(position(labels[i][j],classes))

X_test = []
Y_test = []
for i in range (len(data), len(embeddings)):
  for j in range (len(embeddings[i])):
    X_test.append(embeddings[i][j])
    Y_test.append(position(labels[i][j],classes))
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
import numpy as np
for k in range (len(X_train)):
  X_train[k] = np.array(X_train[k])
```

```python
for k in range (len(X_test)):
  X_test[k] = np.array(X_test[k])
```

```python
# Convertir les données en tensors PyTorch
train_vectors_tensor = torch.tensor(X_train)
train_labels_tensor = torch.tensor(Y_train)
```

```python
# Créer un DataLoader pour les données d'entraînement
train_dataset = TensorDataset(train_vectors_tensor, train_labels_tensor)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
```

```python
# Définir les paramètres
input_dim = len(X_train[0])  # Taille de l'entrée (dimension des vecteurs)
hidden_dim = 128  # Taille de la couche cachée
output_dim = len(classes)  # Nombre de classes de sortie
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# Définition de l'architecture du réseau de neurones feedforward
class FFNNClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(FFNNClassifier, self).__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, output_dim)
        self.softmax = nn.Softmax(dim=1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        x = self.softmax(x)
        return x
```

```python
# Initialiser le modèle
model = FFNNClassifier(input_dim, hidden_dim, output_dim)
```

```python
# Définir la fonction de perte et l'optimiseur
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
# Entraînement du modèle
num_epochs = 100
for epoch in range(num_epochs):
    model.train()
    for inputs, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(inputs.float())
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
```

```python
# Convertir les données en tensors PyTorch
test_vectors_tensor = torch.tensor(X_test)
test_labels_tensor = torch.tensor(Y_test)
```

```python
# Créer un DataLoader pour les données de test
test_dataset = TensorDataset(test_vectors_tensor, test_labels_tensor)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=True)
```

```python
# Évaluation du modèle
model.eval()
predictions = []
true_labels = []
with torch.no_grad():
    for inputs, labels in test_loader:
        outputs = model(inputs.float())
        _, predicted = torch.max(outputs, 1)
        predictions.extend(predicted.tolist())
        true_labels.extend(labels.tolist())
```

```python
from sklearn.metrics import accuracy_score
# Calcul des performances
accuracy = accuracy_score(true_labels, predictions)
print("Accuracy:", accuracy)
```
